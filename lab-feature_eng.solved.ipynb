There are multiple strategies to handle missing data

- Removing all rows or all columns containing missing data.
- Filling all missing values with a value (mean in continouos or mode in categorical for example).
- Filling all missing values with an algorithm.

For this exercise, because we have such low amount of null values, we will drop rows containing any missing value. 

```python
def clean_null_values(df):
    columns = [col for col in df.columns]
    for col in columns:
        if df[col].dtype=="float64":
            df.loc[
                df[col].isna(),
                col
            ]= df[col].mean()
        else:
            df.loc[
                df[col].isna(),
                col
            ]= df[col].mode().iloc[0]
    return df

            
spaceship = clean_null_values(spaceship)
spaceship.info()
```

- **Cabin** is too granular - transform it in order to obtain {'A', 'B', 'C', 'D', 'E', 'F', 'G', 'T'}

```python
spaceship["Cabin_Cleaned"] = spaceship["Cabin"].str.split('/').str[0]
```

- Drop PassengerId and Name

```python
spaceship.drop(columns=["PassengerId","Name"],axis = 1)
```

- For non-numerical columns, do dummies.

**Perform Train Test Split**

```python
X = pd.get_dummies(spaceship[["HomePlanet","Cabin_Cleaned","Destination"]]).astype(int)
y = spaceship["Transported"].astype(int)
X_train, X_test,y_train, y_test = train_test_split(X,y, random_state=42, test_size=0.3)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)

```

**Model Selection**

In this exercise we will be using **KNN** as our predictive model.

KNN CLASSIFIER

```python
def sqrt_y_test(dependent_variable):
    sqrt_n_y_test = int(math.sqrt(len(y_test)))
    if sqrt_n_y_test % 2 == 0:
        sqrt_n_y_test = sqrt_n_y_test + 1
    else:
        sqrt_n_y_test = sqrt_n_y_test
    return sqrt_n_y_test

n = sqrt_y_test(y_test)

classifier = KNeighborsClassifier(n_neighbors=n)
classifier.fit(X_train,y_train)
y_pred = classifier.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(accuracy)

```
